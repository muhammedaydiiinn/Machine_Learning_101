```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


# bu veride ilk satırlar kolın başlığı değildi
data = pd.read_csv('../../data/sepet.csv', header= None)
data
```

```python
data.info()
```

```python
t = []

# her veriiyi listeye ekliyoruz her satırı ürün bazlı
for i in range(0,7501):
    t.append([str(data.values[i,j]) for j in range(0,20)])
```

```python
# pip install  apyori
```

```python
from apyori import apriori

rules =apriori(t,min_support=0.01,min_confidence=0.2,min_lift=3,max_length=2)
```
