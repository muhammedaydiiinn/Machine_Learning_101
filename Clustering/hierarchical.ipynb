```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Importing the dataset
data = pd.read_csv('../data/musteriler.csv')
data
```

```python
from sklearn.cluster import AgglomerativeClustering
# Selecting the features for clustering
X = data.iloc[:, 3:].values
# Fitting Hierarchical Clustering to the dataset
hierarchical = AgglomerativeClustering(n_clusters=4,metric="euclidean", linkage='ward')
Y_tahmin = hierarchical.fit_predict(X)
print(Y_tahmin)
```

```python
plt.scatter(X[Y_tahmin==0,0], X[Y_tahmin==0,1] ,s=100, c='red', label='Cluster 1')
plt.scatter(X[Y_tahmin==1,0], X[Y_tahmin==1,1] ,s=100, c='blue', label='Cluster 2')
plt.scatter(X[Y_tahmin==2,0], X[Y_tahmin==2,1] ,s=100, c='green', label='Cluster 3')
plt.scatter(X[Y_tahmin==3,0], X[Y_tahmin==3,1] ,s=100, c='cyan', label='Cluster 4')
plt.title('Hierarchical Clustering')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
dendrogram = sch.dendrogram(sch.linkage(X, method='ward'))
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean distances')
plt.show()
```
