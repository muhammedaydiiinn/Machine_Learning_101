```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 10000
d =10
oduller = [0]*d
tiklamalar = [0]*d
toplam = 0
chooesed= []
df = pd.read_csv('Ads_CTR_Optimisation.csv')
```

```python
for n in range(0,N):
    ad = 0
    max_ucb = 0
    for i in range(0,d):
        if (tiklamalar[i] > 0):
            ortalama = oduller[i] / (tiklamalar[i])
            delta = np.sqrt(3/2 * np.log(n+1) / (tiklamalar[i]))
            ucb = ortalama + delta
        else:
            ucb = N*10
        if max_ucb < ucb:
            max_ucb = ucb
            ad = i
    chooesed.append(ad)
    tiklamalar[ad] = tiklamalar[ad] + 1
    odul =df.values[n, ad]
    oduller[ad] = oduller[ad] + odul
    toplam += odul
print("Toplam Ödül: ", toplam)
```

```python
plt.hist(chooesed)
plt.show()
```
