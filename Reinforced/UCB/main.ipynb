```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Ads_CTR_Optimisation.csv')
df.head()
```

```python
import random
N = 10000
d =10
total = 0
chooesed= []
for n in range(0, N):
    ad = random.randrange(d)
    chooesed.append(ad)
    prize = df.values[n, ad] # n. satır =1 ise ödül 1, 0 ise ödül 0
    total += prize

print("Toplam Ödül: ", total)
```

```python
plt.hist(chooesed)
plt.title("Seçilen Reklamların Dağılımı")
plt.xlabel("Reklam Numarası")
plt.ylabel("Reklamın Seçilme Sayısı")
plt.show()
```
